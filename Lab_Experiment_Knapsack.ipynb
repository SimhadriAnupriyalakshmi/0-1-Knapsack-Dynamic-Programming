```python
n = 4
wt = [2, 3, 4, 5]
val = [3, 4, 5, 6]
W = 5

dp = [[0]*(W+1) for _ in range(n+1)]

for i in range(1, n+1):
    for w in range(W+1):
        if wt[i-1] <= w:
            dp[i][w] = max(val[i-1] + dp[i-1][w-wt[i-1]],
                            dp[i-1][w])
        else:
            dp[i][w] = dp[i-1][w]

print("Maximum usefulness:", dp[n][W])

w = W
items = []
for i in range(n, 0, -1):
    if dp[i][w] != dp[i-1][w]:
        items.append(i)
        w -= wt[i-1]

print("Selected items:", items[::-1])

print("DP Table:")
for row in dp:
    print(row)
```
